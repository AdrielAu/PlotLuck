# NN2 pretrained CPU inference — human LinuxONE operator only
Run the existing image unchanged. This is inference, not training or NNPA acceleration. Use synthetic cases only. Review benchmark labels before tuning; reserve heldout cases. Keep weights in the external Hugging Face cache, never commit them.


```python
from pathlib import Path
from datetime import date
from plotluck.nn2 import evaluate
root = Path.cwd().resolve()
while not (root / 'pyproject.toml').exists():
    if root == root.parent:
        raise RuntimeError('Run inside the PlotLuck clone')
    root = root.parent
output = root / 'runs' / f'{date.today().isoformat()}-nn2-evaluation' / 'operator-results.json'
report = evaluate(root, output, split='development', limit=3)
print({key: value for key, value in report.items() if key != 'results'})
print([(r['id'], r['seconds']) for r in report['results']])
```

Copy the synthetic report to a laptop and run `node scripts/score-nn2.mjs runs/<yyyy-mm-dd>-nn2-evaluation/operator-results.json`. Archive the report, score, environment and cold/warm timings in a dated run. The report does not establish the 100-case quality gate; run heldout only after human label review and frozen tuning. CPU completion may exceed 60 seconds: record the failure, do not weaken the gate.
